# 01 — Exploratory Data Analysis (MovieLens 1M)

This notebook performs a comprehensive exploratory data analysis of the **MovieLens 1M dataset** from GroupLens.
We investigate:
1. Dataset structure across `users.dat`, `movies.dat`, and `ratings.dat`.
2. Rating distribution and statistics.
3. User demographic distributions (Age, Gender, Occupation).
4. Movie genres distribution.
5. User-item interaction matrix sparsity.


```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

# Add project root to sys.path
ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.data_loader import load_all_data
from src.preprocessing import preprocess_movies, preprocess_ratings

users_df, movies_df, ratings_df = load_all_data()
movies_clean = preprocess_movies(movies_df)
ratings_clean = preprocess_ratings(ratings_df)

print(f"Users shape:   {users_df.shape}")
print(f"Movies shape:  {movies_clean.shape}")
print(f"Ratings shape: {ratings_clean.shape}")
```

## 1. Rating Distribution & Summary Statistics


```python
print(ratings_clean['Rating'].describe())
print("\nRating counts by value:")
print(ratings_clean['Rating'].value_counts().sort_index(ascending=False))
```

## 2. Matrix Sparsity Analysis
Let's compute the sparsity of the User-Movie rating matrix:
$$\text{Sparsity} = 1 - \frac{|\text{Ratings}|}{|\text{Users}| \times |\text{Movies}|}$$


```python
n_users = len(users_df)
n_movies = len(movies_clean)
n_ratings = len(ratings_clean)
total_possible = n_users * n_movies
sparsity_pct = (1.0 - (n_ratings / total_possible)) * 100.0

print(f"Total Users:              {n_users:,}")
print(f"Total Movies:             {n_movies:,}")
print(f"Total Ratings:            {n_ratings:,}")
print(f"Possible Combinations:    {total_possible:,}")
print(f"Matrix Sparsity:          {sparsity_pct:.2f}%")
```

## 3. Movie Genres Distribution


```python
all_genres = []
for g_str in movies_df["Genres"]:
    all_genres.extend(g_str.split("|"))

genre_counts = pd.Series(all_genres).value_counts()
print("Genre Frequencies across 3,883 movies:")
print(genre_counts)
```

## 4. Release Year Distribution


```python
print("Release Year Summary:")
print(movies_clean['Year'].describe())
print("\nTop 10 Release Years with most movies:")
print(movies_clean['Year'].value_counts().head(10))
```
