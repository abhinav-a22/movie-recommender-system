# 02 — Version 1: Basic Content-Based Recommendation

This notebook explores the implementation and behavior of **Version 1: Basic Content-Based Recommendation**.
- **Features:** Movie genres multi-hot vectorization using `CountVectorizer(binary=True)`.
- **Similarity Metric:** Cosine similarity.
- **Rule:** Filter out the selected query movie.


```python
import sys
from pathlib import Path
import pandas as pd

ROOT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

from src.data_loader import load_movies
from src.preprocessing import preprocess_movies
from src.version1.basic_content import BasicContentRecommender

movies_df = load_movies()
movies_clean = preprocess_movies(movies_df)

# Fit Version 1
v1_model = BasicContentRecommender()
v1_model.fit(movies_clean)
print("Vocabulary (Genres):", v1_model.vectorizer.get_feature_names_out())
print("Genre Matrix Shape:", v1_model.genre_matrix.shape)
```

## Recommendations for Toy Story (MovieID = 1)


```python
recs = v1_model.recommend(movie_id=1, top_k=10)
recs_df = pd.DataFrame(recs)
recs_df[['rank', 'title', 'genres', 'similarity_score']]
```
